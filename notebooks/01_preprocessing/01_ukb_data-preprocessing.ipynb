# Raw Data Preprocessing (University of Bonn - Single Channel)

## Import Python's Library

```python
import numpy as np                          # read data library
import matplotlib.pyplot as plt             # visualization
import os                                   # call os function
from scipy.signal import butter, filtfilt   # butterworth signal filtering
```

## Load 90 Data

```python
## Inisialisasi Variabel dan Path Dataset

# empty array
set_a = []  # healthy
set_b = []  # healthy
set_c = []  # interictal
set_d = []  # interictal
set_e = []  # ictal

data_set_a_path = '../../data/ukb_raw/Z'           # healthy/non-epileptic (eyes open) 
data_set_b_path = '../../data/ukb_raw/O'           # healthy/non-epileptic (eyes closed) 
data_set_c_path = '../../data/ukb_raw/N'           # epileptic (interictal)
data_set_d_path = '../../data/ukb_raw/F'           # epileptic (interictal)
data_set_e_path = '../../data/ukb_raw/S'           # epileptic (ictal)

## Load Sinyal EEG Setiap Set 1-90
# set a (dataset Z) 
for i in range(1,91):
    if i < 10:
        filename = 'Z00{}.txt'.format(i)
    elif i < 100:
        filename = 'Z0{}.txt'.format(i)
    else:
        filename = 'Z{}.txt'.format(i)
    signal_a = np.loadtxt(os.path.join(data_set_a_path, filename))
    set_a.append(signal_a)  

# set b (dataset O) 
for i in range(1,91):
    if i < 10:
        filename = 'O00{}.txt'.format(i)
    elif i < 100:
        filename = 'O0{}.txt'.format(i)
    else:
        filename = 'O{}.txt'.format(i)
    signal_b = np.loadtxt(os.path.join(data_set_b_path, filename))
    set_b.append(signal_b)  
        
# set c (dataset N) 
for i in range(1,91):
    if i < 10:
        filename = 'N00{}.txt'.format(i)
    elif i < 100:
        filename = 'N0{}.txt'.format(i)
    else:
        filename = 'N{}.txt'.format(i)
    signal_c = np.loadtxt(os.path.join(data_set_c_path, filename))
    set_c.append(signal_c) 

# set d (dataset F) 
for i in range(1,91):
    if i < 10:
        filename = 'F00{}.txt'.format(i)
    elif i < 100:
        filename = 'F0{}.txt'.format(i)
    else:
        filename = 'F{}.txt'.format(i)
    signal_d = np.loadtxt(os.path.join(data_set_d_path, filename))
    set_d.append(signal_d)
        
# set e (dataset S) 
for i in range(1,91):
    if i < 10:
        filename = 'S00{}.txt'.format(i)
    elif i < 100:
        filename = 'S0{}.txt'.format(i)
    else:
        filename = 'S{}.txt'.format(i)
    signal_e = np.loadtxt(os.path.join(data_set_e_path, filename))
    set_e.append(signal_e)    

print(len(set_a))  # showing the first array
print(set_b)  # showing the first array
print(set_c)  # showing the first array
print(set_d)  # showing the first array
print(set_e)  # showing the first array
```

## Load All Data

```python
## Inisialisasi Variabel dan Path Dataset

# empty array
all_a = []  # healthy
all_b = []  # healthy
all_c = []  # interictal
all_d = []  # interictal
all_e = []  # ictal

data_all_a_path = '../../data/ukb_raw/Z'           # healthy/non-epileptic (eyes open) 
data_all_b_path = '../../data/ukb_raw/O'           # healthy/non-epileptic (eyes closed) 
data_all_c_path = '../../data/ukb_raw/N'           # epileptic (interictal)
data_all_d_path = '../../data/ukb_raw/F'           # epileptic (interictal)
data_all_e_path = '../../data/ukb_raw/S'           # epileptic (ictal)

## Load Sinyal EEG Setiap Set 1-80
# set a (dataset Z) 
for i in range(1,101):
    if i < 10:
        filename = 'Z00{}.txt'.format(i)
    elif i < 100:
        filename = 'Z0{}.txt'.format(i)
    else:
        filename = 'Z{}.txt'.format(i)
    signal_a = np.loadtxt(os.path.join(data_all_a_path, filename))
    all_a.append(signal_a)  

# set b (dataset O) 
for i in range(1,101):
    if i < 10:
        filename = 'O00{}.txt'.format(i)
    elif i < 100:
        filename = 'O0{}.txt'.format(i)
    else:
        filename = 'O{}.txt'.format(i)
    signal_b = np.loadtxt(os.path.join(data_all_b_path, filename))
    all_b.append(signal_b)  
        
# set c (dataset N) 
for i in range(1,101):
    if i < 10:
        filename = 'N00{}.txt'.format(i)
    elif i < 100:
        filename = 'N0{}.txt'.format(i)
    else:
        filename = 'N{}.txt'.format(i)
    signal_c = np.loadtxt(os.path.join(data_all_c_path, filename))
    all_c.append(signal_c) 

# set d (dataset F) 
for i in range(1,101):
    if i < 10:
        filename = 'F00{}.txt'.format(i)
    elif i < 100:
        filename = 'F0{}.txt'.format(i)
    else:
        filename = 'F{}.txt'.format(i)
    signal_d = np.loadtxt(os.path.join(data_all_d_path, filename))
    all_d.append(signal_d)
        
# set e (dataset S) 
for i in range(1,101):
    if i < 10:
        filename = 'S00{}.txt'.format(i)
    elif i < 100:
        filename = 'S0{}.txt'.format(i)
    else:
        filename = 'S{}.txt'.format(i)
    signal_e = np.loadtxt(os.path.join(data_all_e_path, filename))
    all_e.append(signal_e)    

print(len(all_a))  # showing the first array
print(all_b)  # showing the first array
print(all_c)  # showing the first array
print(all_d)  # showing the first array
print(all_e)  # showing the first array
```

## Data Visualization

## Signal Augmentation for Ictal

```python
# augmentation params
time_shift_range = 50 
noise_level = 0.1  
scaling_range = [0.9, 1.1]

# empty array data augmentation
augmented_signal_e = []  
    
# set e augmentation
for signal in set_e:
    # time shifting
    time_shift = int(np.random.uniform(-time_shift_range, time_shift_range))
    if time_shift > 0:
        augmented_signal = np.concatenate((np.zeros((time_shift)), signal[:-time_shift]))
    else:
        augmented_signal = np.concatenate((signal[-time_shift:], np.zeros((-time_shift))))
    augmented_signal = augmented_signal[:4097]
    
    # adding noise
    noise = np.random.normal(0, noise_level, size=augmented_signal.shape)
    augmented_signal = augmented_signal + noise
    
    # scaling
    scaling_factor = np.random.uniform(scaling_range[0], scaling_range[1])
    augmented_signal = augmented_signal * scaling_factor
    
    augmented_signal_e.append(augmented_signal)

# combine augmented signals and seizure signals
augmented_ictal_signal = set_e + augmented_signal_e 

print(len(augmented_ictal_signal))
```

## Signal Filtering (OPSIONAL)

```python
# filter params
lowcut = 0.1
highcut = 80
fs = 173.61
order = 6

# set nyquist frequency
nyquist = 0.5 * fs

low = lowcut / nyquist
high = highcut / nyquist

b, a = butter(order, [low, high], btype='band')


# filtering all normal data
healthy_all = all_a + all_b
interictal_all = all_c + all_d
ictal_all = all_e

# filtering 90% normal data
healthy_signal = set_a + set_b 
interictal_signal = set_c + set_d  
ictal_signal = set_e

# filtering all filtering data
filtered_healthy_all = filtfilt(b, a, healthy_all)
filtered_interictal_all = filtfilt(b, a, interictal_all)
filtered_ictal_all = filtfilt(b, a, ictal_all)

# filtering 90% filtering data
filtered_healthy_signal = filtfilt(b, a, healthy_signal)
filtered_interictal_signal = filtfilt(b, a, interictal_signal)
filtered_ictal_signal = filtfilt(b, a, ictal_signal)

# print len normal data
print("Jumlah Sinyal Sehat :",len(healthy_all))
print("Jumlah Sinyal Interictal :",len(interictal_all))
print("Jumlah Sinyal Ictal :",len(ictal_all),"\n")

print("Jumlah Data Train Sinyal Sehat :",len(healthy_signal))
print("Jumlah Data Train Sinyal Interictal :",len(interictal_signal))
print("Jumlah Data Train Sinyal Ictal :",len(ictal_signal))
print("Jumlah Data Train Sinyal Augmented Ictal :",len(augmented_ictal_signal),"\n")

# print len filtered data
print("Jumlah Semua Sinyal Sehat Terfilter :",len(filtered_healthy_all))
print("Jumlah Semua Sinyal Interictal Terfilter :",len(filtered_interictal_all))
print("Jumlah Semua Sinyal Ictal Terfilter :",len(filtered_ictal_all),"\n")
print("Jumlah Data Train Sehat Terfilter :",len(filtered_healthy_signal))
print("Jumlah Data Train Interictal Terfilter :",len(filtered_interictal_signal))
print("Jumlah Data Train Ictal Terfilter :",len(filtered_ictal_signal))
```
