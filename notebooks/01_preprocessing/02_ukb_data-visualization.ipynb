# Data Visualization (University of Bonn - Single Channel)

## Import Python's Library

```python
import numpy as np                          # read data library
import matplotlib.pyplot as plt             # visualization
```

## Load Data

```python
# load normal data
healthy = np.load("../../data/ukb_processed/normal/healthy_signal.npy")
interictal = np.load("../../data/ukb_processed/normal/interictal_signal.npy")
ictal = np.load("../../data/ukb_processed/normal/augmented_ictal_signal.npy")

print(len(healthy))
print(len(interictal))
print(len(ictal))
```

## Signal Visualization


```python
fs = 173.61  # sampling rate Bonn dataset
time = np.arange(4097) / fs  # waktu dalam detik (0 - 23.6 detik)

# healthy data visualization
fig, axes = plt.subplots(5, 1, figsize=(20, 10))  # 5 baris, 1 kolom
for i in range(5):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, healthy[i])
    axes[i].set_title(f'Original Healthy Signal {i+1}')
    axes[i].set_xlabel('Time')
    axes[i].set_ylabel('Intensity')
    
plt.tight_layout()
plt.show()

# interictal data visualization
fig, axes = plt.subplots(5, 1, figsize=(20, 10))  # 5 baris, 1 kolom
for i in range(5):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, interictal[i])
    axes[i].set_title(f'Original Interictal Signal {i+1}')
    axes[i].set_xlabel('Time')
    axes[i].set_ylabel('Intensity')
    
plt.tight_layout()
plt.show()

# ictal data visualization
fig, axes = plt.subplots(5, 1, figsize=(20, 10))  # 5 baris, 1 kolom
for i in range(5):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, ictal[i])
    axes[i].set_title(f'Original Ictal Signal {i+1}')
    axes[i].set_xlabel('Time')
    axes[i].set_ylabel('Intensity')
    
plt.tight_layout()
plt.show()
```

```python
fixed_ylim = (-1400, 1400)

# healthy data visualization
fig, axes = plt.subplots(3, 1, figsize=(10, 6))  
for i in range(3):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, healthy[i])
    axes[i].set_title(f'Original Healthy Signal {i+1}')
    axes[i].set_xlabel('Time (s)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)  
    
    max_val = np.max(healthy[i])
    min_val = np.min(healthy[i])
    print(f"[Healthy Signal {i+1}] Max: {max_val:.1f}, Min: {min_val:.1f}")
    
plt.tight_layout()
plt.show()

# interictal data visualization
fig, axes = plt.subplots(3, 1, figsize=(10, 6))  
for i in range(3):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, interictal[i])
    axes[i].set_title(f'Original Interictal Signal {i+1}')
    axes[i].set_xlabel('Time (s)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)  
    
    max_val = np.max(interictal[i])
    min_val = np.min(interictal[i])
    print(f"[Interictal Signal {i+1}] Max: {max_val:.1f}, Min: {min_val:.1f}")
    
plt.tight_layout()
plt.show()

# ictal data visualization
fig, axes = plt.subplots(3, 1, figsize=(10, 6))  
for i in range(3):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time, ictal[i])
    axes[i].set_title(f'Original Ictal Signal {i+1}')
    axes[i].set_xlabel('Time (s)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)
    
    
    max_val = np.max(ictal[i])
    min_val = np.min(ictal[i])
    print(f"[Ictal Signal {i+1}] Max: {max_val:.1f}, Min: {min_val:.1f}")
    
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 173.61  
N = int(fs * 5) 
time_ms = np.arange(N) / fs * 1000  

fixed_ylim = (-1400, 1400)

# 1️⃣ Healthy data visualization
fig, axes = plt.subplots(2, 1, figsize=(20, 6))
for i in range(2):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time_ms, healthy[i][:N])
    axes[i].set_title(f'Original Healthy Signal {i+1} (5s)')
    axes[i].set_xlabel('Time (ms)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)
    
plt.tight_layout()
plt.show()

# 2️⃣ Interictal data visualization
fig, axes = plt.subplots(2, 1, figsize=(20, 6))
for i in range(2):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time_ms, interictal[i][:N])
    axes[i].set_title(f'Original Interictal Signal {i+1} (5s)')
    axes[i].set_xlabel('Time (ms)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)
    
plt.tight_layout()
plt.show()

# 3️⃣ Ictal data visualization
fig, axes = plt.subplots(2, 1, figsize=(20, 6))
for i in range(2):
    axes[i].axhline(0, color='grey')
    axes[i].plot(time_ms, ictal[i][:N])
    axes[i].set_title(f'Original Ictal Signal {i+1} (5s)')
    axes[i].set_xlabel('Time (ms)')
    axes[i].set_ylabel('Amplitude (uV)')
    axes[i].set_ylim(fixed_ylim)
    
plt.tight_layout()
plt.show()
```
